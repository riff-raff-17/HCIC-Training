## 7.1 Defining and Calling Functions

A function definition has this structure:

```python
def function_name(parameter1, parameter2):
    # body
    return result
```

- `def` tells Python you are defining a function
- The **parameters** are the inputs the function expects (in the brackets)
- The **body** is the indented block that runs when the function is called
- `return` sends a value back to whoever called the function

A function does **nothing** until it is *called*. Defining it just stores the instructions.

````python
# Define a simple function
def greet(name):
    return 'Hello, ' + name + '!'

# Call it — this is what actually runs the code
message = greet('Alice')
print(message)

# Call it again with a different argument
print(greet('Bob'))
````

````python
# A function with two parameters
def add(x, y):
    return x + y


print(add(3, 5))  # 8
print(add(10, -2))  # 8
````

### Parameters vs Arguments

These two words are often used interchangeably, but they mean slightly different things:

- **Parameters** are the variable names in the function **definition**: `def add(x, y)` — `x` and `y` are parameters
- **Arguments** are the actual values passed in when the function is **called**: `add(3, 5)` — `3` and `5` are arguments

When you call `add(3, 5)`, Python assigns `x = 3` and `y = 5` inside the function.

````python
# Tracing parameter assignment
def describe(animal, sound):
    return 'A ' + animal + ' goes ' + sound

# When this is called:
# animal = 'cat', sound = 'meow'
print(describe('cat', 'meow'))

# animal = 'dog', sound = 'woof'
print(describe('dog', 'woof'))
````

## 7.2 Default Parameter Values

You can give a parameter a **default value** in the function definition. If the caller doesn't provide that argument, the default is used instead.

```python
def f(x, y=10):
    return x + y

f(5)       # y uses default → 5 + 10 = 15
f(5, 3)    # y is overridden → 5 + 3 = 8
```

This appeared directly in **HCIC 2021 Q16**: `def f1(one=1, two=1):`

The question asked:
- (a) What are the **arguments** (the HCIC paper used this word to mean *parameters*)
- (b) What is the output when called with specific values
- (c) What is the output when called with NO values (defaults apply)

````python
# HCIC 2021 Q16 — the exact function from the test
def f1(one=1, two=1):
    return two, one

# The parameters are: one (default=1) and two (default=1)
# Note: it returns two FIRST, then one — ORDER MATTERS!
````

````python
# (b) What is the output for:
a = 5
b = 7
a, b = f1(a, b)
print('a:', a, 'b:', b)

# Trace:
# f1(a, b) → f1(5, 7)
# one=5, two=7
# returns two, one → returns 7, 5
# a, b = 7, 5
# print: a: 7 b: 5
````

````python
# (c) What is the output for:
c, d = f1()
print('c:', c, 'd:', d)

# Trace:
# f1() — no arguments provided, so both defaults apply
# one=1, two=1
# returns two, one → returns 1, 1
# c, d = 1, 1
````

## 7.3 Return Values

The `return` statement sends a value back to the caller and **immediately exits the function**. Code after `return` is never reached.

```python
def check(x):
    if x > 0:
        return 'positive'
    return 'not positive'   # only reached if x <= 0
```

````python
# return exits immediately — code after return doesn't run
def check(x):
    if x > 0:
        return 'positive'     # exits here if x > 0
    return 'not positive'     # only reached if x <= 0

print(check(5))    # 'positive'
print(check(-3))   # 'not positive'
print(check(0))    # 'not positive'
````

### Functions without `return`

If a function has no `return` statement, it returns `None` automatically.

````python
def say_hello():
    print('Hello!')    # prints as a side effect, doesn't return anything

result = say_hello()
print(result)   # None — because no return statement
````

## 7.4 Returning Multiple Values

Python functions can return **more than one value** by separating them with commas. The values come back as a **tuple**.

```python
def min_max(lst):
    return min(lst), max(lst)

lo, hi = min_max([3, 1, 7, 2])
# lo = 1, hi = 7
```

This appeared in **HCIC 2021 Q16**: `return two, one` — note the **reversed** order!

The critical trap: the function returns `two` first and `one` second. So `a, b = f1(a, b)` assigns the old `b` to `a` and the old `a` to `b` — it's a swap!

````python
# Multiple return values come back as a tuple
def split_name(full_name):
    parts = full_name.split(' ')   # splits on the space
    return parts[0], parts[1]     # returns first, last

first, last = split_name('Ada Lovelace')
print('First:', first)
print('Last:', last)
````

````python
# The EXACT function from HCIC 2021 Q16
def f1(one=1, two=1):
    return two, one   # returns TWO first, then ONE

# Calling with a=5, b=7:
a = 5
b = 7
a, b = f1(a, b)
# f1(5, 7): one=5, two=7 → returns (7, 5)
# a, b = (7, 5) → a=7, b=5
print('a:', a, 'b:', b)
````

## 7.5 Variable Scope

Variables created **inside** a function only exist within that function. They are called **local** variables and cannot be accessed from outside.

Variables created **outside** any function are **global** and can be read from anywhere, but you need the `global` keyword to modify them inside a function.

This concept is less directly tested on HCIC but is important for understanding why tracing function calls works the way it does.

````python
# Local variables only exist inside the function
def my_func():
    local_var = 100  # this only exists inside my_func
    return local_var


result = my_func()
print(result)  # 100 — the returned value

# Trying to access local_var here would cause an error:
# print(local_var)   # NameError!
````

````python
# Parameters are also local to the function
def double(x):
    x = x * 2  # this changes the LOCAL x only
    return x


n = 5
result = double(n)
print("result:", result)  # 10
print("n is still:", n)  # 5 — unchanged! n is not x
````

## 7.6 Keyword Arguments

When calling a function, you can specify arguments by **name** instead of just by position. This is useful when a function has default parameters and you only want to override one of them.

```python
def greet(name, greeting='Hello'):
    return greeting + ', ' + name

greet('Alice')                    # uses default greeting
greet('Alice', 'Hi')              # overrides by position
greet(greeting='Hey', name='Bob') # overrides by name — order doesn't matter
```

````python
def describe_pet(name, animal="cat"):
    return name + " is a " + animal


print(describe_pet("Whiskers"))  # uses default
print(describe_pet("Rex", "dog"))  # positional
print(describe_pet(animal="fish", name="Nemo"))  # keyword — order swapped!
````

## 7.7 Tracing Function Calls Step by Step

When tracing a function call, follow these steps every time:

1. **Identify the arguments** in the call
2. **Map arguments to parameters** in the definition (left to right, unless named)
3. **Apply defaults** for any parameters not provided
4. **Run the function body** with those parameter values
5. **Identify the return value** and substitute it back at the call site

Let's practise this methodically.

````python
# Step-by-step tracing example
def compute(a, b, c=10):
    result = a + b
    if result > c:
        return result * 2
    return result + c

# Call 1: compute(3, 4)
# Step 1: arguments = 3, 4
# Step 2: a=3, b=4
# Step 3: c=10 (default)
# Step 4: result = 3+4 = 7. 7 > 10? No → return 7+10
# Step 5: returns 17
print(compute(3, 4))

# Call 2: compute(6, 7)
# Step 1: arguments = 6, 7
# Step 2: a=6, b=7
# Step 3: c=10 (default)
# Step 4: result = 6+7 = 13. 13 > 10? Yes → return 13*2
# Step 5: returns 26
print(compute(6, 7))

# Call 3: compute(3, 4, 5)
# Step 3: c=5 (overrides default)
# Step 4: result = 7. 7 > 5? Yes → return 7*2
print(compute(3, 4, 5))
````

## 7.8 Object-Oriented Programming Basics

Object-Oriented Programming (OOP) organises code around **objects** — bundles of data and behaviour. This appeared in **HCIC 2021 Q17**.

You only need to understand the basics for the test:

| Term | What it is |
|------|------------|
| **Class** | A blueprint/template for creating objects |
| **Object / Instance** | A specific thing created from a class |
| **Attribute** | A variable that belongs to an object (`self.name`) |
| **Method** | A function that belongs to a class |
| `__init__` | Special method — runs automatically when an object is created |
| `self` | Refers to the current object inside a method |

````python
# HCIC 2021 Q17 — the exact class from the test
class Cat:
    def __init__(self, name, age):
        self.name = name  # stores name as an attribute
        self.age = age  # stores age as an attribute

    def sound(self):
        return "Meow..."
````

To create an object from a class, call the class like a function:

### Extending the class — adding more methods

Let's practise reading a slightly more complex class definition.

````python
class Robot:
    def __init__(self, name, battery=100):
        self.name = name
        self.battery = battery
        self.is_on = False

    def turn_on(self):
        self.is_on = True
        return self.name + ' is now ON'

    def move(self, steps):
        if self.is_on:
            self.battery -= steps
            return self.name + ' moved ' + str(steps) + ' steps'
        return 'Robot is off!'

    def status(self):
        return self.name + ': battery=' + str(self.battery)
````

```python
def encode(message, shift=1):
    result = ''
    for char in message:
        if char.isalpha():
            result += chr(ord(char) + shift)
        else:
            result += char
    return result

print(encode('abc'))        # 'bcd'
print(encode('abc', 2))     # 'cde'
print(encode('Hello!', 1))  # 'Ifmmp!'
```

## 📝 Key Takeaways

**1. `def` defines. Calling runs.**  
Defining a function stores the instructions. Nothing actually runs until you call it. Don't confuse `def greet(name):` with `greet('Alice')`.

**2. Parameters get values from arguments, left to right.**  
`def f(a, b)` + `f(3, 5)` → `a=3`, `b=5`. Always map them in order.

**3. Default parameters are used when the argument is omitted.**  
`def f(x, y=10)` — if you call `f(5)`, then `y=10`. If you call `f(5, 3)`, then `y=3`.

**4. Return order matters when returning multiple values.**  
`return two, one` returns `two` first. If you do `a, b = f(a, b)`, make sure you know which value lands in which variable.

**5. Local variables don't exist outside the function.**  
Parameters and variables created inside a function are local. Changing them doesn't affect variables outside the function with the same name.

**6. For OOP: a class is a blueprint, an instance is the object.**  
`__init__` runs when you create an instance. `self` refers to the instance itself. Attributes are accessed with `object.attribute`. Methods are called with `object.method()`.

**7. The HCIC 2021 function question in full:**  
`def f1(one=1, two=1): return two, one` — parameters are `one` and `two`, both defaulting to 1. It swaps and returns them. Called with no args → `(1, 1)`. Called with `(5, 7)` → `(7, 5)`.
