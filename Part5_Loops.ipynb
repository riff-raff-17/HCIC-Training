## 5.1 `for` Loops — Iterating Over a Sequence

A `for` loop runs its body once for each item in a sequence:

```python
for variable in sequence:
    # body — runs once per item
```

The `variable` takes the value of each item in turn. The loop ends when all items have been visited.

````python
# Looping over a list
for item in [10, 20, 30, 40]:
    print(item)
````

````python
# Looping over a string — visits one character at a time
for char in "HCIC":
    print(char)
````

````python
# The loop variable can be used inside the body
for n in [1, 2, 3, 4, 5]:
    print(n, "squared is", n * n)
````

## 5.2 `range()` — Generating Number Sequences

`range()` is the most common way to control how many times a `for` loop runs. It generates a sequence of integers.

There are three forms:

| Form | What it produces |
|------|------------------|
| `range(n)` | 0, 1, 2, ..., n-1 |
| `range(start, stop)` | start, start+1, ..., stop-1 |
| `range(start, stop, step)` | start, start+step, ..., up to but not including stop |

**The golden rule:** `range()` always stops **before** the `stop` value. `range(5)` gives 0–4, NOT 0–5.

````python
# range(n) — from 0 up to but NOT including n
print("range(5):", list(range(5)))  # 0, 1, 2, 3, 4
print("range(3):", list(range(3)))  # 0, 1, 2
````

````python
# range(start, stop) — from start up to but NOT including stop
print("range(2, 7):", list(range(2, 7)))  # 2, 3, 4, 5, 6
print("range(1, 4):", list(range(1, 4)))  # 1, 2, 3
````

````python
# range(start, stop, step) — every 'step' values
print("range(0, 10, 2):", list(range(0, 10, 2)))  # 0, 2, 4, 6, 8
print("range(1, 10, 3):", list(range(1, 10, 3)))  # 1, 4, 7
print("range(10, 0, -1):", list(range(10, 0, -1)))  # 10, 9, 8, ..., 1 (countdown!)
````

### Using `range()` in a `for` loop

````python
# Run something 4 times
for i in range(4):
    print("Pass", i)
````

````python
# Use i as an index into a list
fruits = ["apple", "banana", "cherry"]
for i in range(len(fruits)):
    print(i, fruits[i])
````

## 5.3 `while` Loops — Repeat Until False

A `while` loop keeps running as long as its condition is `True`:

```python
while condition:
    # body — runs as long as condition is True
```

The condition is checked **before** each pass. When it becomes `False`, the loop stops and execution continues after the loop.

**Important:** If the condition never becomes `False`, you get an **infinite loop**. Always make sure something inside the loop moves you toward the exit condition.

````python
# Basic while loop — count from 1 to 5
n = 1
while n <= 5:
    print(n)
    n += 1  # without this, n stays at 1 forever!
````

````python
# While loop — keep halving until below 1
x = 100
while x >= 1:
    print(x)
    x //= 2
````

### `for` vs `while` — when to use which?

- Use `for` when you know **how many times** you want to loop, or when iterating over a fixed sequence
- Use `while` when you loop until a **condition** is met, and you don't know in advance how many passes it will take

On the HCIC test, `while` loops are used almost exclusively for **tracing questions** — you're given the code and must work out what the variable values are after the loop ends.

## 5.4 Loop Tracing — The Core Exam Skill

The most important skill for HCIC loop questions is being able to **trace through a loop step by step** and track the value of every variable after each pass.

### The Tracing Table Method

For any loop question, draw a table with:
- One column per variable
- One row per pass of the loop
- A final row for after the loop ends

Let's practise with a real HCIC example.

```python
i, x = 0, 5
while x > i:
    i += 1
    x -= 1
result = i + x
print(result)
```

**Tracing table:**

| Pass | i (start) | x (start) | x > i? | i (end) | x (end) |
|------|-----------|-----------|--------|---------|---------|
| 1    | 0         | 5         | Yes    | 1       | 4       |
| 2    | 1         | 4         | Yes    | 2       | 3       |
| 3    | 2         | 3         | Yes    | 3       | 2       |
| 4    | 3         | 2         | **No** | —       | —       |

Loop exits with i=3, x=2. result = 3+2 = **5**.

````python
# Verify the trace
i, x = 0, 5
while x > i:
    i += 1
    x -= 1
result = i + x
print(result)
````

### Adding print statements to see inside a loop

When tracing is hard, add temporary `print()` calls inside the loop to see what's happening at each pass. This is a debugging technique, not just a study trick.

````python
# Same loop but with tracing prints added
i, x = 0, 5
pass_num = 0
while x > i:
    pass_num += 1
    print(f"Pass {pass_num}: i={i}, x={x}, x>i={x>i}")
    i += 1
    x -= 1
    print(f"  After: i={i}, x={x}")

print(f"Loop ended: i={i}, x={x}")
print("result =", i + x)
````

## 5.5 `break` and `continue`

These two keywords give you more control over loop flow:

- `break` — **immediately exits** the loop, skipping any remaining iterations
- `continue` — **skips the rest of the current pass** and jumps to the next iteration

These appear less frequently on HCIC but are worth knowing.

````python
# break — exits the loop early
for n in range(10):
    if n == 5:
        break  # stop the loop when n reaches 5
    print(n)
print("Loop ended")
````

````python
# continue — skips the rest of this pass only
for n in range(10):
    if n % 2 == 0:
        continue  # skip even numbers
    print(n)  # only odd numbers print
````

## 5.6 Combining Loops with Conditionals

The most common HCIC pattern: a loop iterates over a sequence, and a conditional inside the loop decides what to do with each item. We saw this with strings in Part 3. Now let's see it with numbers and lists.

The general pattern:

```python
accumulator = starting_value
for item in sequence:
    if condition:
        update accumulator
print(accumulator)
```

````python
# Count how many numbers in a list are even
numbers = [1, 4, 7, 2, 9, 6, 3, 8]
count = 0
for n in numbers:
    if n % 2 == 0:
        count += 1
print('Even count:', count)
````

````python
# Sum only numbers greater than 5
numbers = [1, 4, 7, 2, 9, 6, 3, 8]
total = 0
for n in numbers:
    if n > 5:
        total += n
print('Sum of numbers > 5:', total)
````

## 5.7 The `while T_or_F` Pattern

HCIC 2023 Q14 introduced a subtle pattern: using a variable as the while condition directly, relying on **truthiness**.

```python
T_or_F = 0
while T_or_F:
    ...
```

Since `T_or_F = 0` and `0` is **falsy**, the loop body **never runs at all**. The output is whatever the accumulator was initialised to.

````python
# HCIC 2023 Q14 — the while condition is 0 (falsy)
lst = [1, 'a, b', 4, 'five', [6]]
T_or_F = 0
count = 0
while T_or_F:
    for item in lst:
        count += 1
print(count)
````

````python
# Contrast: what if T_or_F = 1 (truthy)?
# WARNING: this would be an infinite loop! We cap it at 1 pass to demo:
lst = [1, 'a, b', 4, 'five', [6]]
T_or_F = 1
count = 0
passes = 0
while T_or_F and passes < 1:   # safety cap
    for item in lst:
        count += 1
    passes += 1
print('After 1 pass, count =', count)
````

## 5.8 Nested Loops

A loop inside another loop is called a **nested loop**. The inner loop runs completely for each pass of the outer loop.

This appeared in **HCIC 2023 Q14** and similar tracing questions.

````python
# Basic nested loop
for i in range(3):          # outer loop: 0, 1, 2
    for j in range(2):      # inner loop: 0, 1  (runs fully each time)
        print(f'i={i}, j={j}')
    print('--- end of outer pass', i)
````

### Counting total iterations

If the outer loop runs **m** times and the inner loop runs **n** times, the body of the inner loop runs **m × n** times total.

````python
# Count total iterations
count = 0
for i in range(4):      # 4 passes
    for j in range(3):  # 3 passes each time
        count += 1
print('Total iterations:', count)  # 4 × 3 = 12
````

## 📝 Key Takeaways

**1. `for` loops iterate over a sequence. `while` loops run until a condition is False.**  
Use `for` when you know how many iterations. Use `while` when you're waiting for a condition to be met.

**2. `range()` stops BEFORE the stop value.**  
`range(5)` gives 0, 1, 2, 3, 4 — NOT 5. `range(1, 6)` gives 1, 2, 3, 4, 5 — NOT 6. This off-by-one error is the most common loop mistake.

**3. Always trace loops with a table.**  
Write out every variable's value after every pass. This is the only reliable way to solve loop questions under exam pressure.

**4. The while condition is checked BEFORE each pass.**  
If the condition is already False at the start (like `while 0:`), the loop body never runs at all.

**5. Truthiness applies to while conditions.**  
`while 0:` never runs. `while '':` never runs. `while 'hello':` runs until `'hello'` becomes `''`.

**6. Watch for variables changing INSIDE the loop.**  
When `a = a * 2` and `b -= a` are both in the same pass, the order matters — `b -= a` uses the ALREADY UPDATED value of `a`.

**7. `break` exits the loop immediately. `continue` skips the rest of the current pass.**  
These are different! `break` ends the loop entirely. `continue` just skips the rest of this one iteration.
