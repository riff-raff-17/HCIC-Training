## 6.1 Creating Lists and Accessing Elements

A list is an **ordered, changeable** collection of items. It's written with square brackets `[]` and items separated by commas.

```python
my_list = [10, 20, 30, 40, 50]
```

Like strings, lists use **zero-based indexing** — the first item is at index 0.

```
Value:  10   20   30   40   50
Index:   0    1    2    3    4
Neg:    -5   -4   -3   -2   -1
```

````python
scores = [85, 72, 90, 61, 78]

print(scores[0])  # first element
print(scores[2])  # third element
print(scores[-1])  # last element
print(scores[-2])  # second to last
````

````python
# len() gives the number of items
print(len(scores))  # 5
````

````python
# Slicing works exactly like strings
print(scores[1:4])  # [72, 90, 61]
print(scores[:3])  # [85, 72, 90]
print(scores[::2])  # [85, 90, 78] — every other element
````

## 6.2 Modifying List Elements

Unlike strings, **lists are mutable** — you can change individual elements after creation.

```python
lst[index] = new_value
```

This is a critically important difference from strings. If you try `'hello'[0] = 'H'` you get an error. But with lists, `lst[0] = 'H'` works perfectly.

````python
grades = ["C", "B", "A", "D", "B"]
print("Before:", grades)

grades[3] = "A"  # change index 3 from 'D' to 'A'
print("After: ", grades)
````

````python
# Strings are IMMUTABLE — this would cause an error:
# word = 'hello'
# word[0] = 'H'   # TypeError!

# Lists are MUTABLE — this works fine:
letters = ["h", "e", "l", "l", "o"]
letters[0] = "H"
print(letters)
````

## 6.4 Two Ways to Loop Over a List

There are two common patterns for looping over a list. The HCIC test uses both, and it's crucial to know **which one to use when**.

### Pattern 1: `for item in list` — when you just need the values

````python
# Use this when you only need to READ each value
scores = [85, 72, 90, 61, 78]
total = 0
for score in scores:
    total += score
print('Total:', total)
````

### Pattern 2: `for i in range(len(list))` — when you need the index

Use this when you need to **modify elements** by position, or when you need to know which position you're at.

````python
# Use this when you need to MODIFY elements by index
scores = [85, 72, 90, 61, 78]
for i in range(len(scores)):
    if scores[i] < 75:
        scores[i] = 'Fail'
print(scores)
````

### ⚠️ Why you can't modify items with Pattern 1

With `for item in list`, `item` is a **copy** of the value. Changing `item` does NOT change the original list.

````python
# This does NOT change the list!
scores = [85, 72, 90, 61, 78]
for score in scores:
    score = score + 10   # this only changes the local copy
print(scores)           # unchanged!
````

````python
# This DOES change the list — use range(len())
scores = [85, 72, 90, 61, 78]
for i in range(len(scores)):
    scores[i] = scores[i] + 10
print(scores)           # all scores +10
````

## 6.5 Modifying Lists Inside Loops — The Core HCIC Pattern

This is the most tested list pattern on HCIC. The structure is always:

```python
lst = [original values]
for i in range(len(lst)):
    if condition involving lst[i]:
        lst[i] = new_value
    else:
        lst[i] = other_value
print(lst)
```

The question either asks:
1. **What is the output?** (trace it)
2. **What is the condition?** (work backwards from the expected output)

Both appeared on **HCIC 2023 B4** and **HCIC 2024 B4**.

````python
# HCIC 2023 B4 — predict the condition
# Output should be ['Fail', 'Pass', 'Fail', 'Pass', 'Fail']
test = [25, 50, 49, 80, 49.5]
for i in range(5):
    if test[i] < 50:    # condition: score strictly less than 50
        test[i] = 'Fail'
    else:
        test[i] = 'Pass'
print(test)
````

````python
# HCIC 2024 B4 — predict the condition
# Output should be ['Distinction', 'Pass', 'Pass', 'Distinction', 'Pass']
results = ['A1', 'B3', 'C5', 'A2', 'B4']
for i in range(5):
    if results[i][0] == 'A':    # condition: first character is 'A'
        results[i] = 'Distinction'
    else:
        results[i] = 'Pass'
print(results)
````

## 6.6 Mixed-Type Lists

Python lists can hold items of **different types** in the same list — integers, strings, other lists, floats, all mixed together.

This appeared in **HCIC 2023 Q14**: `[1, 'a, b', 4, 'five', [6]]`

A nested list (a list inside a list) counts as **one element**.

````python
mixed = [1, 'hello', 3.14, True, [10, 20]]

print(len(mixed))    # 5 — the inner list counts as ONE item
print(mixed[0])      # 1
print(mixed[1])      # 'hello'
print(mixed[4])      # [10, 20]
print(mixed[4][0])   # 10 — first element of the nested list
````

## 6.7 Tuples — Like Lists but Immutable

A **tuple** is like a list but with one critical difference: **tuples cannot be changed** after creation. They are *immutable*.

- List: uses square brackets `[1, 2, 3]`
- Tuple: uses parentheses `(1, 2, 3)`

Everything you know about list indexing and slicing works exactly the same with tuples.

````python
# Creating a tuple
t = (10, 20, 30, 40, 50)

print(t[0])      # 10 — indexing works the same
print(t[-1])     # 50
print(t[1:4])    # (20, 30, 40) — slicing works too
print(len(t))    # 5
````

````python
# Tuples are IMMUTABLE — you cannot change elements
# Uncomment to see the error:

# t = (10, 20, 30)
# t[0] = 99    # TypeError: 'tuple' object does not support item assignment
````

### Tuple concatenation with `+`

Even though you can't modify a tuple, you can create a **new** tuple by concatenating two tuples with `+`.

This is exactly what `+=` does with tuples — it creates a new tuple and reassigns the variable.

This appeared in **HCIC 2022 A15**.

````python
# Concatenation creates a NEW tuple
a = (1, 2, 3)
b = (4, 5)
c = a + b
print(c)    # (1, 2, 3, 4, 5)
````

````python
# += with tuples — creates a new tuple and reassigns
word = ('H', 'C')
print('Before:', word)

word += ('I', 'C')    # same as word = word + ('I', 'C')
print('After: ', word)
````

## 6.8 List and Tuple Summary Table

| Feature | List | Tuple |
|---------|------|-------|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` |
| Mutable? | ✅ Yes — can change elements | ❌ No — fixed after creation |
| Indexing | `lst[0]` | `tpl[0]` |
| Slicing | `lst[1:3]` | `tpl[1:3]` |
| `len()` | ✅ Yes | ✅ Yes |
| `in` | ✅ Yes | ✅ Yes |
| `+` concatenation | Creates new list | Creates new tuple |
| `.append()` | ✅ Yes | ❌ No |
| Can modify with `lst[i] = x` | ✅ Yes | ❌ Error |

## 📝 Key Takeaways

**1. Lists use square brackets `[]`. Tuples use parentheses `()`.**  
Both support indexing, slicing, `len()`, and `in`. The critical difference is mutability.

**2. Lists are mutable — tuples are not.**  
`lst[0] = 99` works fine. `tpl[0] = 99` causes a TypeError.

**3. To modify elements, loop with `range(len(lst))` not `for item in lst`.**  
`for item in lst` gives you copies. Only `lst[i] = new_value` actually changes the list.

**4. Nested lists count as one element.**  
`[1, [2, 3], 4]` has length 3, not 4. Access the inner list with `lst[1]` and its elements with `lst[1][0]`.

**5. The modify-in-place loop is the most tested list pattern.**  
Loop with `range(len())`, check a condition on `lst[i]`, assign a new value back to `lst[i]`. Know how to trace this and how to reverse-engineer the condition from the output.

**6. `.append()` adds to the end of a list. `+` creates a new list.**  
`lst.append(x)` modifies the existing list. `lst + [x]` creates a brand new list without changing the original.

**7. Tuple `+=` creates a new tuple, not an error.**  
`tpl += (4, 5)` is valid — it creates a new tuple and reassigns `tpl`. The original tuple object is not modified.
